```python
# Employee payroll systtem

employee_ids=[]

employees={}

def calculate_allowance(basic_salary):
    return basic_salary * 0.20

def calculate_deduction(basic_salary):
    return basic_salary * 0.10

class Employee:
     def __init__(self, emp_id, name, basic_salary):
        self.emp_id = emp_id
        self.name = name
        self.basic_salary = basic_salary

     def calculate_salary(self):
        allowance = calculate_allowance(self.basic_salary)
        deduction = calculate_deduction(self.basic_salary)
        net_salary = self.basic_salary + allowance - deduction
        return allowance, deduction, net_salary

n = int(input("Enter no of employeees: "))
for i in range(n):
    try:
         emp_id = input("\nEnter Employee ID: ")
         name = input("Enter Employee Name: ")


         basic_salary = float(input("Enter Basic Salary: "))


         if basic_salary < 0:

             raise ValueError("Salary cannot be negative!")



         employee_ids.append(emp_id)

         employees[emp_id] = Employee(emp_id, name, basic_salary)
    except ValueError as e:
         print("Invalid Salary Entry:", e)


print("\n" + "=" * 60)
print("EMPLOYEE PAYROLL REPORT")
print("=" * 60)

for emp_id in employee_ids:
    emp = employees[emp_id]

    allowance, deduction, net_salary = emp.calculate_salary()

    print(f"\nEmployee ID   : {emp.emp_id}")
    print(f"Employee Name : {emp.name}")
    print(f"Basic Salary  : ₹{emp.basic_salary:.2f}")
    print(f"Allowance     : ₹{allowance:.2f}")
    print(f"Deduction     : ₹{deduction:.2f}")
    print(f"Net Salary    : ₹{net_salary:.2f}")

print("\nTotal Employees:", len(employee_ids))
```
