```python
# Numpy statistics Dashboard(CLI)
import numpy as np

numbers = []

while True:
    print("\n===== NumPy Statistics Dashboard =====")
    print("1. Enter Numbers")
    print("2. Display Numbers")
    print("3. Mean")
    print("4. Median")
    print("5. Standard Deviation")
    print("6. Variance")
    print("7. Maximum")
    print("8. Minimum")
    print("9. Sum")
    print("10. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        data = input("Enter numbers separated by space: ")
        numbers = np.array(list(map(float, data.split())))
        print("Data Stored Successfully!")

    elif choice == "2":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Numbers:", numbers)

    elif choice == "3":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Mean =", np.mean(numbers))

    elif choice == "4":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Median =", np.median(numbers))

    elif choice == "5":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Standard Deviation =", np.std(numbers))

    elif choice == "6":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Variance =", np.var(numbers))

    elif choice == "7":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Maximum =", np.max(numbers))

    elif choice == "8":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Minimum =", np.min(numbers))

    elif choice == "9":
        if len(numbers) == 0:
            print("No data available.")
        else:
            print("Sum =", np.sum(numbers))

    elif choice == "10":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
