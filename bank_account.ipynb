```python
#bank account system
accounts = {}
acc_numbers = []

class BankAccount:
    def __init__(self, acc_no, name, balance):
        self.acc_no = acc_no
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        self.save_transaction(f"Deposited ₹{amount}")

    def withdraw(self, amount):
        if amount > self.balance:
            raise Exception("Insufficient Balance!")
        self.balance -= amount
        self.save_transaction(f"Withdrawn ₹{amount}")

    def check_balance(self):
        return self.balance

    def save_transaction(self, msg):
        try:
            with open("transactions.txt", "a") as file:
                file.write(f"Account {self.acc_no}: {msg}\n")
        except PermissionError:
            print("Cannot access transaction file.")

acc_no = input("Enter Account Number: ")
name = input("Enter Name: ")
balance = float(input("Enter Initial Balance: "))

accounts[acc_no] = BankAccount(acc_no, name, balance)
acc_numbers.append(acc_no)

deposit_amount = float(input("Enter Deposit Amount: "))
accounts[acc_no].deposit(deposit_amount)

try:
    amount = float(input("Enter Withdrawal Amount: "))
    accounts[acc_no].withdraw(amount)
except Exception as e:
    print(e)

print("Current Balance:", accounts[acc_no].check_balance())
```
