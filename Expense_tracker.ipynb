```python
def add_expense():
    try:
        amount = float(input("Enter amount: "))
        category = input("category: ")


        with open("expense.txt","a") as f:
            f.write(f"{category},{amount}\n")

        print("Expense added")

    except ValueError:

        print("Invalid Amount")

add_expense()
```
