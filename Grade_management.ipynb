```python
import csv

# Create sample students.csv
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Marks"])
    writer.writerow(["Shubh", 95])
    writer.writerow(["Aman", 80])
    writer.writerow(["Riya", 45])

# Function to calculate grade
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"

# Read students.csv and create result.csv
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)

    with open("result.csv", "w", newline="") as out:
        writer = csv.writer(out)
        writer.writerow(["Name", "Marks", "Grade"])

        for row in reader:
            writer.writerow([row[0], row[1], grade(int(row[1]))])

print("Results saved in result.csv")

# Display result.csv
with open("result.csv", "r") as file:
    print(file.read())
```
